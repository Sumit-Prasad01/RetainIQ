```python
import warnings
warnings.filterwarnings("ignore")
import joblib
from scipy.stats import randint, uniform
from sklearn.model_selection import RandomizedSearchCV
from lightgbm import LGBMClassifier
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score, accuracy_score
```

```python
from constants import TRAIN_DATA_PATH
```

```python
data = joblib.load(f"../../{TRAIN_DATA_PATH}")

X_train = data['X_train']
y_train = data['y_train']
X_test = data['X_test']
y_test = data['y_test']
```

```python
# Declare parameters
param_grid = {
    'n_estimators' : randint(100, 500),
    "max_depth" : randint(5, 50),
    "learning_rate" : uniform(0.01, 0.2),
    'num_leaves' : randint(20, 100),
    'boosting_type' : ['gbdt', 'dart', 'goss']
}
```

```python
LIGHTGBM_PARAMS = {
    'n_estimators': randint(100, 500),
    'max_depth': randint(5, 50),
    'learning_rate': uniform(0.01, 0.2),
    'num_leaves': randint(20, 100),
    'boosting_type': ['gbdt', 'dart', 'goss']
}
```

```python
LIGHTGBM_EXTENSIVE_PARAMS = {
    "n_estimators": randint(200, 1000),
    "learning_rate": uniform(0.01, 0.15),
    "num_leaves": randint(20, 150),
    "max_depth": randint(3, 20),
    "min_child_samples": randint(10, 100),
    "subsample": uniform(0.6, 0.4),
    "colsample_bytree": uniform(0.6, 0.4),
    "reg_alpha": uniform(0, 2),
    "reg_lambda": uniform(0, 2),
    "boosting_type": ["gbdt", "dart"]
}
```

```python
lgbm_grid = RandomizedSearchCV(
    estimator=LGBMClassifier(
        objective="binary",
        random_state=42
    ),
    param_distributions=LIGHTGBM_PARAMS,
    n_iter=10,
    cv=5,
    n_jobs=-1,
    verbose=0,
    random_state=42,
    scoring="accuracy"
)
```

```python
lgbm_random = RandomizedSearchCV(
    estimator=LGBMClassifier(
        objective="binary",
        random_state=42,
        n_jobs=-1
    ),
    param_distributions=LIGHTGBM_EXTENSIVE_PARAMS,
    n_iter=50,
    cv=5,
    scoring="f1",
    n_jobs=-1,
    verbose=0,
    random_state=42
)
```

```python
lgbm_random.fit(X_train , y_train)
best_model = lgbm_random.best_estimator_
best_model
```

```python
def evaluate_model(model, X_test, y_test):
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:,1]
    
    return {
        "Accuracy Score": accuracy_score(y_test, y_pred),
        "Recall_Churn": recall_score(y_test, y_pred),
        "Precision_Churn": precision_score(y_test, y_pred),
        "F1_Churn": f1_score(y_test, y_pred),
        "ROC_AUC": roc_auc_score(y_test, y_prob)
    }
```

```python
evaluate_model(best_model, X_test, y_test)
```

```python
y_prob = lgbm_random.predict_proba(X_test)[:, 1]

for threshold in [0.5, 0.45, 0.40, 0.35, 0.30, 0.25]:
    y_pred = (y_prob >= threshold).astype(int)

    print(
        threshold,
        precision_score(y_test, y_pred),
        recall_score(y_test, y_pred),
        f1_score(y_test, y_pred)
    )
```

```python
import numpy as np
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

thresholds = np.arange(0.10, 0.91, 0.01)

best_threshold = 0
best_f1 = 0

for threshold in thresholds:

    y_pred = (y_prob >= threshold).astype(int)

    f1 = f1_score(y_test, y_pred)

    if f1 > best_f1:
        best_f1 = f1
        best_threshold = threshold

print("Best Threshold:", best_threshold)
print("Best F1:", best_f1)
```

```python
BEST_THRESHOLD = best_threshold

y_pred_final = (y_prob >= BEST_THRESHOLD).astype(int)

final_metrics = {
    "Accuracy Score": accuracy_score(y_test, y_pred_final),
    "Recall_Churn": recall_score(y_test, y_pred_final),
    "Precision_Churn": precision_score(y_test, y_pred_final),
    "F1_Churn": f1_score(y_test, y_pred_final),
    "ROC_AUC": roc_auc_score(y_test, y_prob)
}

print(final_metrics)
```
